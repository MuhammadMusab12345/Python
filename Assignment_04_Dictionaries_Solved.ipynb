Q1 — Basic Dictionary Creation
Create a dictionary called student with keys: name, age, course, city. Print each value.

```python
student = {
    "name": "Ali",
    "age": 20,
    "course": "Computer Science",
    "city": "Karachi"
}

print("Name:", student["name"])
print("Age:", student["age"])
print("Course:", student["course"])
print("City:", student["city"])
```

Q2 — Adding & Updating Values
car = {"brand": "Toyota", "model": "Corolla", "year": 2018}
Add color='White', update year to 2022 and print dictionary.

```python
car = {
    "brand": "Toyota",
    "model": "Corolla",
    "year": 2018
}

# Adding a new key-value pair
car["color"] = "White"

# Updating an existing value
car["year"] = 2022

print("Updated dictionary:", car)
```

Q3 — Loop Through Dictionary
person = {"name": "Ali", "age": 25, "city": "Karachi"}
Print keys and values using loop.

```python
person = {
    "name": "Ali",
    "age": 25,
    "city": "Karachi"
}

for key, value in person.items():
    print(key, ":", value)
```

Q4 — Count Frequency of Numbers
numbers = [1,2,2,3,3,3,4,4,4,4]
Create frequency dictionary.

```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}
for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

print("Frequency dictionary:", frequency)
```

Q5 — Word Counter
Take sentence input and count word frequency.

```python
sentence = input("Enter a sentence: ")
words = sentence.split()

word_count = {}
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print("Word frequency:", word_count)
```

Q6 — Student Marks & Average
marks = {"Math":80, "English":70, "Physics":90}
Calculate total and average.

```python
marks = {"Math": 80, "English": 70, "Physics": 90}

total = sum(marks.values())
average = total / len(marks)

print("Total marks:", total)
print("Average marks:", average)
```

Q7 — Highest Scoring Student
students = {"Ali":85, "Sara":92, "Ahmed":78, "Zara":95}
Find highest scoring student.

```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest_name = None
highest_score = -1

for name, score in students.items():
    if score > highest_score:
        highest_score = score
        highest_name = name

print("Highest scoring student is:", highest_name, "with score:", highest_score)
```

Q8 — Merge Dictionaries
dict1 = {"a":1,"b":2}
dict2 = {"c":3,"d":4}
Merge and print result.

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}

merged_dict = {**dict1, **dict2}

print("Merged dictionary:", merged_dict)
```

Q9 — Mini Project: Contact Book
Add 3 contacts, search by name, print phone or Contact not found.

```python
contacts = {}

# Add 3 contacts
for i in range(3):
    name = input("Enter contact name: ")
    phone = input("Enter phone number: ")
    contacts[name] = phone

print("\nContact book:", contacts)

# Search by name
search_name = input("\nEnter name to search: ")

if search_name in contacts:
    print("Phone number:", contacts[search_name])
else:
    print("Contact not found")
```
