# List Assignment

1. Write a program that accepts a list from user and print the alternate element of list.

```python
lst = list(map(int, input("Enter list elements separated by space: ").split()))

print("Alternate elements of the list are:")
for i in range(0, len(lst), 2):
    print(lst[i], end=" ")
```

2. Write a program that accepts a list from user. Your program should reverse the content of list and display it. Do not use reverse() method.

```python
lst = list(map(int, input("Enter list elements separated by space: ").split()))

reversed_list = []
for i in range(len(lst) - 1, -1, -1):
    reversed_list.append(lst[i])

print("Reversed list:", reversed_list)
```

3. Find and display the largest number of a list without using built-in function max(). Your program should ask the user to input values in list from keyboard.

```python
lst = list(map(int, input("Enter list elements separated by space: ").split()))

largest = lst[0]
for num in lst:
    if num > largest:
        largest = num

print("Largest number in the list is:", largest)
```

4. Write a program that rotates the element of a list so that the element at the first index moves to the second index, the element in the second index moves to the third index, etc., and the element in the last index moves to the first index.

```python
lst = list(map(int, input("Enter list elements separated by space: ").split()))

# Element at last index goes to first index, everyone else shifts one place forward
last_element = lst[-1]
rotated_list = [last_element] + lst[:-1]

print("Original list:", lst)
print("Rotated list:", rotated_list)
```

5. Write a program that input a string and ask user to delete a given word from a string.

```python
sentence = input("Enter a string: ")
word = input("Enter the word to delete: ")

words = sentence.split()
new_words = [w for w in words if w != word]
new_sentence = " ".join(new_words)

print("String after deleting the word:", new_sentence)
```

6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It should print the date in the form March 12, 2021.

```python
date_str = input("Enter date in mm/dd/yyyy format: ")

month, day, year = date_str.split("/")

months = ["January", "February", "March", "April", "May", "June",
          "July", "August", "September", "October", "November", "December"]

month_name = months[int(month) - 1]

print(f"{month_name} {int(day)}, {year}")
```

7. Write a program with a function that accepts a string from keyboard and create a new string after converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses." the output should be "Stop And Smell The Roses"

```python
def capitalize_words(sentence):
    words = sentence.split()
    new_words = []
    for word in words:
        new_word = word[0].upper() + word[1:]
        new_words.append(new_word)
    return " ".join(new_words)

sentence = input("Enter a sentence: ")
result = capitalize_words(sentence)
print("Capitalized sentence:", result)
```

8. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be like this :

2  11  7  12
5  2   9  15
8  3   10 42

Sum of row 1 = 32
Sum of row 2 = 31
Sum of row 3 = 63

```python
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

matrix = []
print("Enter the matrix row by row (space separated values):")
for i in range(rows):
    row = list(map(int, input(f"Row {i+1}: ").split()))
    matrix.append(row)

for i in range(rows):
    row_sum = sum(matrix[i])
    print(f"Sum of row {i+1} = {row_sum}")
```

9. Write a program to add two matrices of size n x m.

```python
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

print("Enter elements of first matrix:")
matrix1 = []
for i in range(rows):
    row = list(map(int, input(f"Row {i+1}: ").split()))
    matrix1.append(row)

print("Enter elements of second matrix:")
matrix2 = []
for i in range(rows):
    row = list(map(int, input(f"Row {i+1}: ").split()))
    matrix2.append(row)

result = []
for i in range(rows):
    result_row = []
    for j in range(cols):
        result_row.append(matrix1[i][j] + matrix2[i][j])
    result.append(result_row)

print("Sum of the two matrices is:")
for row in result:
    print(row)
```

10. Write a program to multiply two matrices

```python
r1 = int(input("Enter rows of first matrix: "))
c1 = int(input("Enter columns of first matrix: "))
r2 = int(input("Enter rows of second matrix: "))
c2 = int(input("Enter columns of second matrix: "))

if c1 != r2:
    print("Matrix multiplication not possible (columns of first must equal rows of second).")
else:
    print("Enter elements of first matrix:")
    matrix1 = []
    for i in range(r1):
        row = list(map(int, input(f"Row {i+1}: ").split()))
        matrix1.append(row)

    print("Enter elements of second matrix:")
    matrix2 = []
    for i in range(r2):
        row = list(map(int, input(f"Row {i+1}: ").split()))
        matrix2.append(row)

    # Initialize result matrix with zeros
    result = [[0 for _ in range(c2)] for _ in range(r1)]

    for i in range(r1):
        for j in range(c2):
            for k in range(c1):
                result[i][j] += matrix1[i][k] * matrix2[k][j]

    print("Product of the two matrices is:")
    for row in result:
        print(row)
```
