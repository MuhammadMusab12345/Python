**QUESTION 1: Working with Lists**

1. Create a list called numbers containing at least 8 integers.
2. Using a for loop, do the following:
   - Print all numbers in the list.
   - Print the sum of all numbers.
   - Print only the even numbers.

```python
# 1. Create a list of at least 8 integers
numbers = [12, 45, 7, 23, 9, 56, 34, 18]

# 2. Print all numbers using a for loop
print("All numbers in the list:")
total = 0
for num in numbers:
    print(num, end=" ")
    total += num  # keep adding to total for the sum

# Print the sum of all numbers
print("\n\nSum of all numbers:", total)

# Print only the even numbers
print("\nEven numbers in the list:")
for num in numbers:
    if num % 2 == 0:
        print(num, end=" ")
```

**QUESTION 2: Dictionary Basics**

1. Create a dictionary called student with the following keys: name, age, course, marks (must store a list of 3 marks)
2. Using loops:
   - Print all keys and their values.
   - Calculate the average of the marks.
   - Display whether the student Passed (average >= 50) or Failed.

```python
# 1. Dictionary with a list inside it (marks)
student = {
    "name": "Ali",
    "age": 21,
    "course": "Computer Science",
    "marks": [70, 85, 60]
}

# Print all keys and values using a loop
for key, value in student.items():
    print(key, ":", value)

# Calculate the average of marks using a loop
total_marks = 0
for mark in student["marks"]:
    total_marks += mark

average = total_marks / len(student["marks"])
print("\nAverage marks:", average)

# Display Passed or Failed
if average >= 50:
    print("Result: Passed")
else:
    print("Result: Failed")
```

**QUESTION 3: List of Dictionaries**

1. Create a list called employees.
2. Each employee must be stored as a dictionary containing: name, department, salary
3. Using loops:
   - Print details of all employees.
   - Find the employee with the highest salary.
   - Calculate the total salary expense.

```python
# 1 & 2. List of dictionaries (minimum 3 employees)
employees = [
    {"name": "Ali", "department": "IT", "salary": 50000},
    {"name": "Sara", "department": "HR", "salary": 45000},
    {"name": "Ahmed", "department": "Finance", "salary": 60000}
]

# Print details of all employees using a loop
print("Employee Details:")
for emp in employees:
    print(f"Name: {emp['name']}, Department: {emp['department']}, Salary: {emp['salary']}")

# Find the employee with the highest salary (comparison logic)
highest_emp = employees[0]
for emp in employees:
    if emp["salary"] > highest_emp["salary"]:
        highest_emp = emp

print("\nEmployee with highest salary:", highest_emp["name"], "-", highest_emp["salary"])

# Calculate total salary expense
total_salary = 0
for emp in employees:
    total_salary += emp["salary"]

print("Total salary expense:", total_salary)
```

**QUESTION 4: While Loop with User Input**

1. Ask the user to enter numbers.
2. Store the numbers in a list.
3. Stop when the user enters -1.
4. After the loop ends:
   - Print the complete list.
   - Print the largest number.
   - Print the smallest number.

```python
numbers_list = []

# While loop that keeps taking input until user enters -1
while True:
    num = int(input("Enter a number (-1 to stop): "))
    if num == -1:
        break
    numbers_list.append(num)

print("\nComplete list:", numbers_list)

if numbers_list:
    largest = numbers_list[0]
    smallest = numbers_list[0]
    for num in numbers_list:
        if num > largest:
            largest = num
        if num < smallest:
            smallest = num
    print("Largest number:", largest)
    print("Smallest number:", smallest)
else:
    print("No numbers were entered.")
```

**QUESTION 5: Word Frequency Counter**

1. Ask the user to enter a sentence.
2. Convert the sentence into a list of words.
3. Use a dictionary to count how many times each word appears.
4. Print the word frequency clearly.

Example Format (for understanding only):
apple: 3
banana: 2
orange: 1

```python
sentence = input("Enter a sentence: ")

# Convert sentence into a list of words
words = sentence.split()

# Dictionary used to count frequency of each word
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print("\nWord Frequency:")
for word, count in frequency.items():
    print(f"{word}: {count}")
```

**QUESTION 6: Nested Loops — Multiplication Table**

Using nested loops, print a multiplication table from 1 to 5.

Expected Output Format:
1 2 3 4 5
2 4 6 8 10
3 6 9 12 15
4 8 12 16 20
5 10 15 20 25

```python
# Nested loop: outer loop for rows, inner loop for columns
for i in range(1, 6):
    for j in range(1, 6):
        print(i * j, end=" ")
    print()  # move to next line after each row
```
