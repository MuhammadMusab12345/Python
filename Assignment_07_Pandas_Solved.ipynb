1. Create a DataFrame from a dictionary with columns Name, Age, City and Salary for 6 employees. Print the first 3 rows, the shape, the column names and the data types.

2. Create a Series of 5 numbers with custom index labels ['a','b','c','d','e']. Print the value at label 'c', the values at positions 1 to 3, and the sum and mean of the Series.

6. Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

```python
# Work on a copy so the original df stays unchanged for later questions
df6 = df.copy()

df6["Bonus"] = df6["Salary"] * 0.10
df6["Total"] = df6["Salary"] + df6["Bonus"]
print("After adding Bonus and Total:\n", df6)

df6 = df6.rename(columns={"City": "Location"})
print("\nAfter renaming City to Location:\n", df6)

df6 = df6.drop(columns=["Bonus"])
print("\nAfter dropping Bonus:\n", df6)
```

7. Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
import numpy as np
import pandas as pd

df7 = pd.DataFrame({
    "Name": ["Ali", "Sara", None, "Zara", "Bilal"],
    "Age": [25, np.nan, 30, 28, np.nan],
    "Salary": [40000, 50000, np.nan, 60000, 45000]
})
print("DataFrame with missing values:\n", df7)

# Count of missing values in each column
print("\nMissing values per column:\n", df7.isnull().sum())

# Fill numeric NaNs with column mean
numeric_cols = df7.select_dtypes(include="number").columns
df7[numeric_cols] = df7[numeric_cols].fillna(df7[numeric_cols].mean())
print("\nAfter filling numeric NaNs with mean:\n", df7)

# Drop rows that still have missing values (the missing Name)
df7 = df7.dropna()
print("\nAfter dropping rows with missing values:\n", df7)
```

8. Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
# Sort by Salary (descending), then Age (ascending) for equal salaries
sorted_df = df.sort_values(by=["Salary", "Age"], ascending=[False, True])
print("Sorted DataFrame:\n", sorted_df)

# Top 3 highest-paid employees
print("\nTop 3 highest-paid employees:\n", df.nlargest(3, "Salary"))
```

9. Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
result = df.groupby("City")["Salary"].agg(["mean", "count", "max"])
result.columns = ["Average Salary", "Employee Count", "Max Salary"]

print(result)
```

10. Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
import pandas as pd

employees = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed", "Zara"],
    "DeptID": [1, 2, 3, 5]
})

departments = pd.DataFrame({
    "DeptID": [1, 2, 3, 4],
    "DeptName": ["IT", "HR", "Finance", "Sales"]
})

# Inner join: only matching DeptID in both
print("Inner join:\n", pd.merge(employees, departments, on="DeptID", how="inner"))

# Left join: all employees, NaN where no department matches
print("\nLeft join:\n", pd.merge(employees, departments, on="DeptID", how="left"))

# Concatenate two DataFrames of the same structure
df_a = pd.DataFrame({"Name": ["Ali", "Sara"], "DeptID": [1, 2]})
df_b = pd.DataFrame({"Name": ["Sara", "Ahmed"], "DeptID": [2, 3]})

combined = pd.concat([df_a, df_b], ignore_index=True)
print("\nAfter concat:\n", combined)

# Remove duplicate rows
print("\nAfter removing duplicates:\n", combined.drop_duplicates().reset_index(drop=True))
```
