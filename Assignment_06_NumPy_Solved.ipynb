1. Create a 1D NumPy array of integers from 10 to 50 (inclusive, step 5). Print the array, its shape, dtype and number of dimensions.

```python
import numpy as np

# np.arange(start, stop, step): stop is exclusive, so use 51 to include 50
arr = np.arange(10, 51, 5)

print("Array:", arr)
print("Shape:", arr.shape)
print("Dtype:", arr.dtype)
print("Number of dimensions:", arr.ndim)
```

2. Create a 3x4 array of zeros, a 3x4 array of ones, and a 4x4 identity matrix. Then create an array of 8 evenly spaced numbers between 0 and 1 using linspace.

```python
import numpy as np

zeros_arr = np.zeros((3, 4))
ones_arr = np.ones((3, 4))
identity = np.eye(4)                 # 4x4 identity matrix
lin = np.linspace(0, 1, 8)           # 8 evenly spaced numbers from 0 to 1

print("3x4 Zeros:\n", zeros_arr)
print("\n3x4 Ones:\n", ones_arr)
print("\n4x4 Identity matrix:\n", identity)
print("\nLinspace (8 numbers between 0 and 1):\n", lin)
```

3. Create a 1D array of numbers from 1 to 24 and reshape it into a 2D array of shape (4, 6). Then reshape it into a 3D array of shape (2, 3, 4) and print the shape of each.

```python
import numpy as np

arr_1d = np.arange(1, 25)            # numbers 1 to 24
arr_2d = arr_1d.reshape(4, 6)        # 2D array
arr_3d = arr_1d.reshape(2, 3, 4)     # 3D array

print("1D array shape:", arr_1d.shape)
print("2D array shape:", arr_2d.shape)
print("2D array:\n", arr_2d)
print("\n3D array shape:", arr_3d.shape)
print("3D array:\n", arr_3d)
```

4. Given arr = np.array([12, 45, 7, 23, 56, 89, 34, 21]), print: the first 3 elements, the last 3 elements, every alternate element, and the array in reverse order.

```python
import numpy as np

arr = np.array([12, 45, 7, 23, 56, 89, 34, 21])

print("First 3 elements:", arr[:3])
print("Last 3 elements:", arr[-3:])
print("Every alternate element:", arr[::2])
print("Reversed array:", arr[::-1])
```

5. Create a 2D array of shape (5, 5) with values 1 to 25. Extract (a) the second row, (b) the third column, (c) the 3x3 sub-matrix from the center.

```python
import numpy as np

arr = np.arange(1, 26).reshape(5, 5)
print("5x5 array:\n", arr)

# (a) Second row -> index 1
print("\n(a) Second row:", arr[1])

# (b) Third column -> index 2
print("(b) Third column:", arr[:, 2])

# (c) 3x3 sub-matrix from the center -> rows 1 to 3, columns 1 to 3
print("(c) Center 3x3 sub-matrix:\n", arr[1:4, 1:4])
```

6. Given a = np.array([10, 20, 30, 40]) and b = np.array([1, 2, 3, 4]), perform addition, subtraction, multiplication, division and power (a ** 2). Also compute the dot product of a and b.

```python
import numpy as np

a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power (a ** 2):", a ** 2)
print("Dot product:", np.dot(a, b))
```

7. Given arr = np.array([[4, 8, 2], [9, 1, 7], [5, 6, 3]]), find the sum, mean, min, max and standard deviation of the whole array, and then the sum along rows (axis=1) and along columns (axis=0).

```python
import numpy as np

arr = np.array([[4, 8, 2], [9, 1, 7], [5, 6, 3]])

print("Sum:", arr.sum())
print("Mean:", arr.mean())
print("Min:", arr.min())
print("Max:", arr.max())
print("Standard deviation:", arr.std())

# axis=1 -> sum across each row, axis=0 -> sum down each column
print("Sum along rows (axis=1):", arr.sum(axis=1))
print("Sum along columns (axis=0):", arr.sum(axis=0))
```

8. Create an array of 20 random integers between 1 and 100. Using boolean masking, print (a) all elements greater than 50, (b) all even elements, (c) count of elements divisible by 5.

```python
import numpy as np

# randint upper bound is exclusive, so use 101 to include 100
arr = np.random.randint(1, 101, 20)
print("Random array:", arr)

# (a) Elements greater than 50
print("(a) Elements greater than 50:", arr[arr > 50])

# (b) Even elements
print("(b) Even elements:", arr[arr % 2 == 0])

# (c) Count of elements divisible by 5
print("(c) Count of elements divisible by 5:", np.sum(arr % 5 == 0))
```

9. Use np.where to replace all negative values in np.array([5, -3, 8, -1, 0, -7, 4]) with 0. Also find the index of the maximum and minimum value using argmax and argmin.

```python
import numpy as np

arr = np.array([5, -3, 8, -1, 0, -7, 4])

# Replace negatives with 0, keep other values as they are
new_arr = np.where(arr < 0, 0, arr)
print("Original array:", arr)
print("After replacing negatives with 0:", new_arr)

# Index of max and min values (in the original array)
print("Index of maximum value:", np.argmax(arr))
print("Index of minimum value:", np.argmin(arr))
```

10. Create two 2x3 arrays and join them vertically (vstack) and horizontally (hstack). Then sort the array np.array([[9, 3, 6], [2, 8, 1]]) along each row and flatten the result into 1D.

```python
import numpy as np

a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([[7, 8, 9], [10, 11, 12]])

print("Vertical stack (vstack):\n", np.vstack((a, b)))
print("\nHorizontal stack (hstack):\n", np.hstack((a, b)))

# Sort each row, then flatten into 1D
arr = np.array([[9, 3, 6], [2, 8, 1]])
sorted_arr = np.sort(arr, axis=1)
print("\nSorted along each row:\n", sorted_arr)
print("Flattened (1D):", sorted_arr.flatten())
```
